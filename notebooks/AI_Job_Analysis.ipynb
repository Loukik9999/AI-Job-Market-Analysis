```python
import pandas as pd
import numpy as mp
```

```python
df=pd.read_excel("AI Jobs Data.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['posted_date']=pd.to_datetime(df['posted_date'])
```

```python
df.isnull().sum()
```

```python
df["company"] = df["company"].fillna("Unknown Company")
```

```python
df["salary_min"] = df["salary_min"].fillna("NA")
```

```python
df["salary_max"] = df["salary_max"].fillna("NA")
```

```python
df["required_skills"] = df["required_skills"].fillna("Not Specified")
```

```python
df["posted_date"] = df["posted_date"].fillna("No Date")
```

```python
df.isnull().sum()
```

```python
df.to_excel("Clean_AI_Jobs.xlsx",index=False)
```
